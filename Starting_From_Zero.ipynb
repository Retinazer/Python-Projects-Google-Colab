```python
information = {
    "jay" : {
        "phone" : 552,
        "email" : "jay@gmail.com"
    },
    "jeremiah" : {
        "phone" : 555,
        "email" : "jeremiah@gmail.com"
    }

}
user_name = input("What is your name ?")
if user_name in information :
  print(information[user_name]["phone"])
  print(information[user_name]["email"])
else :
  name = input("Who are you ?")
  phone = int(input("What is your phone number ?"))
  email = input("Your email ?")
  information[name] = {
      "phone" : phone,
      "email" : email
  }
```

```python
#Project 2, Report Cleaner and Check Passer
```

```python


def clean_grades(grades) :
  for i in range(len(grades)) :
    if grades[i] == None :
      grades[i] = 0.0

  return grades
def get_average(grades) :
  total = 0
  for x in grades :
    total += x
  return total / len(grades)
def maximum(grades) :
  current_max = -1
  for x in grades :
    if x > current_max :
      current_max = x
  return current_max
def minimum(grades) :
  current_min = 100
  for x in grades :
    if x < current_min :
      current_min = x
  return current_min

raw_grades = [85.5, 92.0, None, 45.0, 78.5, None, 51.0, 30.0]
print(f"The class average is {get_average(clean_grades(raw_grades))}")
print(f"The highest grade is {maximum(clean_grades(raw_grades))}")
print(f"The lowest grade is {minimum(clean_grades(raw_grades))}")
```

```python
def show_item(data) :
  print("CURRENT INVENTORY")
  for row in data:
    name = row[0]
    amount = row[1]
    price = row[2]
    print(f"Name : {name}, Product Amount : {amount}, Product Price : {price}")

def restock_item(data,name,amount) :
  for row in data :
    if row[0] == name :
      row[1] += amount
      return data
  return data

def remove_stockouts(data) :
  for row in data :
    if row[1] <= 0 :
      data.remove(row)
  return data





inventory = [
    ["Apples", 50, 0.50],
    ["Oranges", 20, 0.75],
    ["Bananas", 0, 0.30],
    ["Grapes", 15, 2.00]
]
show_item(inventory)
remove_stockouts(inventory)
restock_item(inventory,"Apples",10)
show_item(inventory)

```
